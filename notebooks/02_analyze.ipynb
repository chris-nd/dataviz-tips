## La largeur de la marge autorisée et intervalle de seuil `[Q1 − marge ; Q3 + marge]`

### 1. Facture

```python
Q1_BILL = tips["total_bill"].quantile(0.25)
Q3_BILL = tips["total_bill"].quantile(0.75)
IQR_BILL = Q3_BILL - Q1_BILL
MARGIN_BILL = 1.5 * IQR_BILL

LIMIT_LEFT_BILL = Q1_BILL - MARGIN_BILL
LIMIT_RIGHT_BILL = Q3_BILL + MARGIN_BILL

print("Seuil inférieur pour les additions:", LIMIT_LEFT_BILL)
print("Seuil supérieur pour les additions:", LIMIT_RIGHT_BILL)
```

```python
print(
    f"Nombre d'observations avec une facture supérieure à {LIMIT_RIGHT_BILL}:",
    tips[tips["total_bill"] > LIMIT_RIGHT_BILL].shape[0],
)

print(
    f"Le pourcentage d'observations avec une facture supérieure à {LIMIT_RIGHT_BILL}:",
    (tips[tips["total_bill"] > LIMIT_RIGHT_BILL].shape[0] / tips["total_bill"].count()),
)

tips[tips["total_bill"] > LIMIT_RIGHT_BILL]
```

#### Constat: 

- Intervalle d'addition: [-2.81, 40.36]

- L'addition max(50,81) est au delà de la borne haute: 9 additions la dépassent

- Le pourcentage d'observations avec une addition supérieure à 40.36 est de 3.703%

- Les factures avec des valeurs extrêmes supérieur à 40.36 n'ont pas toutes des pourboires élevés

### 2. Pourboire

```python
Q1_TIP = tips["tip"].quantile(0.25)
Q3_TIP = tips["tip"].quantile(0.75)
IQR_TIP = Q3_TIP - Q1_TIP
MARGIN_TIP = 1.5 * IQR_TIP

LIMIT_LEFT_TIP = Q1_TIP - MARGIN_TIP
LIMIT_RIGHT_TIP = Q3_TIP + MARGIN_TIP

print("Seuil inférieur pour les pourboires:", LIMIT_LEFT_TIP)
print("Seuil supérieur pour les pourboires:", LIMIT_RIGHT_TIP)
```

```python
print(
    f"Nombre d'observations avec un pourboire supérieur à {LIMIT_RIGHT_TIP}:",
    tips[tips["tip"] > LIMIT_RIGHT_TIP].shape[0],
)

print(
    f"Le pourcentage d'observations avec un pourboire supérieur à {LIMIT_RIGHT_TIP}:",
    (tips[tips["tip"] > LIMIT_RIGHT_TIP].shape[0] / tips["tip"].count()),
)

tips[tips["tip"] > LIMIT_RIGHT_TIP]
```

```python
total_bill = tips[tips["tip"] > LIMIT_RIGHT_TIP]["total_bill"]
mask = total_bill > tips["total_bill"].median()
mask.all()
```

#### Constat: 
- Intervalle de pourboire: [-0.34, 5.93]

- Le pourboire max(10) est au delà de la borne haute: 8 pourboires la dépassent

- Le pourcentage d'observations avec un pourboire supérieur à 5.93 est de 3.292%

- Les pourboires avec des valeurs extrêmes supérieur à 5.93 ont toutes des factures supérieur à la médiane

### 3. Corrélation entre la Facture et le Pourboire

#### Critère retenu:

**Critère retenu** : une corrélation est jugée stable si elle varie de moins de 0,1 en valeur absolue. Ce seuil est une convention choisie avant le calcul, sans autorité statistique universelle.
    

**Étiquettes de corrélation**: convention adaptée de Cohen (1988)

|\|r\|	      |Intensité    |
|-------------|-------------|
|< 0,1	      |négligeable  |
|0,1 à < 0,3  |faible       |
|0,3 à < 0,5  |moyenne      |
|≥ 0,5	      |forte        |

```python
tips[(tips["total_bill"] > LIMIT_RIGHT_BILL) & (tips["tip"] > LIMIT_RIGHT_TIP)]
```

```python
lines = tips[(tips["total_bill"] > LIMIT_RIGHT_BILL) | (tips["tip"] > LIMIT_RIGHT_TIP)]
print(lines.shape[0])
```

```python
bill_tip = tips[
    (tips["total_bill"] <= LIMIT_RIGHT_BILL) & (tips["tip"] <= LIMIT_RIGHT_TIP)
]
print(bill_tip.shape[0])
```

```python
bill_tip.corr(numeric_only=True)
```

#### Constat 1: 

- Nombre d'observations avec une facture supérieure à 40.36 et un pourboire supérieur à 5.93: 3

- Le montant du pourboire est corrélé avec le montant de la facture à 0.605, ce qui indique une relation forte entre les deux variables.

- Le nombre de client par table est corrélé avec le montant de la facture à 0.561, ce qui indique une relation forte entre les deux variables.

- Le montant du pourboire est corrélé avec le nombre de client par table à 0.413, ce qui indique une relation moyenne entre les deux variables.

- Factures extrêmes (> 40,36 ; 9 lignes, 3,703 %) : rares mais rien n'indique qu'elles soient erronées ; toutes n'ont pas un gros pourboire, ce qui est cohérent avec une relation positive forte. On les garde.

- Pourboires extrêmes (> 5,93 ; 8 lignes, 3,292 %) : tous associés à des factures supérieures à la médiane, ce qui est cohérent avec la relation facture-pourboire. On les garde.
