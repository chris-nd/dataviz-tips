## Doublon

```python
tips[tips.duplicated()]
```

**Constat** : 
- Présence d'un doublon sur la ligne 202

**Hypothèse**: 
- doublon probable de saisie, non vérifiable sans identifiant

```python
tips[tips.duplicated(keep=False)]
```

### Décision

- Supprimer le doublon évite d'introduire un biais silencieux même si c'est négligeable avec un unique doublon

```python
tips = tips.drop_duplicates()
```

**Vérification de la suppression**:

```python
tips[tips.duplicated(keep=False)]
```
