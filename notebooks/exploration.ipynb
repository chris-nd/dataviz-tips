## Analyse de corrélation

### Analyse de la corrélation entre le montant de la facture, le pourboire et le sexe des clients

```python
tips[["total_bill", "tip"]].describe()
```

```python
tips.loc[:, ["total_bill", "tip"]].groupby(tips["sex"]).sum()
```

```python
tips.loc[:, ["total_bill", "tip"]].groupby(tips["sex"]).count()
```

```python
tips.loc[:, ["total_bill", "tip"]].groupby(tips["sex"]).mean()
```

### Analyse de la corrélation entre le montant de la facture, le pourboire et une table de fumeurs ou de non fumeurs

```python
tips.loc[:, ["total_bill", "tip"]].groupby(tips["smoker"]).count()
```

```python
tips.loc[:, ["total_bill", "tip"]].groupby(tips["smoker"]).sum()
```

```python
tips.loc[:, ["total_bill", "tip"]].groupby(tips["smoker"]).mean()
```
