# Visualisation exploratoire

```python
import matplotlib.pyplot as plt
import seaborn as sns

from dataviz_tips.data import load_tips
```

```python
tips = load_tips()
```

## Facture

**Question**: 

**Réponse**: 

```python
ax = sns.histplot(tips, x="total_bill")
ax.set_title("Distribution du montant de la facture")
ax.set_xlabel("Montant de la facture")
ax.set_ylabel("Nombre de factures")

plt.show()
```

## Pourboire

**Question**: 

**Réponse**: 

```python
ax = sns.histplot(tips, x="tip")
ax.set_title("Distribution du montant du pourboire")
ax.set_xlabel("Montant du pourboire")
ax.set_ylabel("Nombre de pourboires")

plt.show()
```

## Facture vs Pourboire

**Question**: Le « montant du pourboire » mesure-t-il la générosité, ou la taille de l'addition ?

**Réponse**: Le montant du pourboire est lié au montant de la facture

```python
ax = sns.scatterplot(tips, x="total_bill", y="tip")
ax.set_title("Comparaison du montant de la facture et du pourboire")
ax.set_xlabel("Montant de la facture")
ax.set_ylabel("Pourboire")

plt.show()
```

## Facture vs Valeur seuil

**Question**: Les factures et les valeurs extrêmes correspondent-elles aux bornes calculées?

**Réponse**: Oui, les moustaches et les points isolés correspondent aux bornes calculées.

```python
ax = sns.boxplot(tips, y="total_bill")
ax.set_title("Distribution des factures")
ax.set_xlabel("Factures")
ax.set_ylabel("Montant")

plt.show()
```

## Pourboire vs Valeur seuil

**Question**: Les pourboires et les valeurs extrêmes correspondent-ils aux bornes calculées?

**Réponse**: Oui, les moustaches et les points isolés correspondent aux bornes calculées.

```python
ax = sns.boxplot(tips, y="tip")
ax.set_title("Distribution des pourboires")
ax.set_xlabel("Pourboires")
ax.set_ylabel("Montant")

plt.show()
```

## Size vs Facture

**Question**: Quelle est la répartition des factures par  nombre de personnes à table?

**Réponse**: La majorité des tables sont des tables de 2 personnes (autour de 64%), suivies des tables de 3 et 4 personnes (autour de 15 %). Les tables de 1, 5 et 6 personnes sont rares (≤ 2 %)

```python
ax = sns.countplot(x="size", data=tips, stat="proportion")
ax.set_title("Répartition des factures par nombre de personnes à table")
ax.set_xlabel("Nombre de personnes")
ax.set_ylabel("Proportion de factures")

plt.show()
```
